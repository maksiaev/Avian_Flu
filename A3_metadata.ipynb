```python
# Function to find matches for all segments
def find_A3_matches(elephant_seal_metadata, segment_file):

    a3_west_pb2 = metadata_from_fasta(segment_file)

    # If SRR exists, match based on SRA_runs. Else, match based on File. 

    # sra_metadata = elephant_seal_metadata[elephant_seal_metadata["status"] == "FOUND"]
    elephant_seal_metadata["Accession"] = elephant_seal_metadata["SRA_runs"]
    a3_west_pb2["NVSL_accession"] = a3_west_pb2["Isolate"].apply(lambda x: x.split("-original")[0])

    # print(elephant_seal_metadata)
    # print(a3_west_pb2)

    common_columns = list(a3_west_pb2.columns.intersection(elephant_seal_metadata.columns)) #.difference(["Accession", "NVSL_accession"]))

    print(common_columns)

    a3_west_pb2_sra = a3_west_pb2.merge(elephant_seal_metadata, on="Accession", how="left")
    # print(a3_west_pb2_sra)

    a3_west_pb2_nvsl = a3_west_pb2.merge(elephant_seal_metadata, on="NVSL_accession", how="left")
    # print(a3_west_pb2_nvsl)

    # a3_west_pb2_sra["File"] = a3_west_pb2_sra["Isolate"].apply(lambda x: x.split("-original")[0])
    # a3_west_pb2_sra_gisaid = a3_west_pb2_sra.merge(elephant_seal_metadata, on="File", how="inner")

    a3_west_pb2_common = pd.merge(a3_west_pb2_sra, a3_west_pb2_nvsl, how="outer") #.drop_duplicates() # (keep="first")
    a3_west_pb2_right = a3_west_pb2_common.merge(elephant_seal_metadata, how="left")
    # a3_west_pb2_common = pd.merge(a3_west_pb2_sra, a3_west_pb2_nvsl, how="inner") #.drop_duplicates() # (keep="first")

    return a3_west_pb2_right
```
