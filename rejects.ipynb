```python
downloads = "C:/Users/maksiaevai.NCBI_NT/Documents/Avian_Flu_Files/"

# genotype = "B3.2"
# input_path = downloads + "Combinations/NCBI_Virus_Andersen_GISAID/11-01-2021--12-05-2025_Antarctica_North_America_South_America/" + genotype + "/"
# input_path = downloads + "Combinations/NCBI_Virus_Andersen_GISAID/11-01-2021--09-19-2025_Antarctica_North_America_South_America/"
input_path = downloads + "Paloma/corrected/"

os.chdir(input_path)
reject_lines = []

# output_path = input_path + genotype + "_test.fasta" # Genotype and Segment should all be the same

for dirpath, dirs, files in os.walk(input_path):
    for file in files:
        file_name = os.path.join(dirpath, file)
        print(file_name)
        if "rejects_id" in file_name: # and genotype in file_name:
            with open(file_name) as r:
                lines = r.readlines()
                for i, line in enumerate(lines):
                    reject_lines.append(">" + line.split("|")[0].strip())
                r.close()
    break 

print(reject_lines)
```

```python
for dirpath, dirs, files in os.walk(input_path):
    for file in files:
        file_name = os.path.join(dirpath, file)
        # if genotype in file_name:
        output_file = open(file_name[:-6] + "_edit.fasta", "w")
        with open(file_name) as f:
            lines = f.readlines()
            current_reject = 0
            current_keep = len(lines) - 1
            for l, line in enumerate(lines):
                if line.split("|")[0] in reject_lines:
                    print("rejected head:", file_name, line)
                    current_reject = l
                    print("current reject:", l)
                elif ">" in line and line not in reject_lines: # This is not being caught before we have to get rid of sequence line
                    output_file.write(line)
                    current_keep = l 
                    # print("current keep:", l)
                elif l > current_reject and ">" not in line and current_keep < current_reject:
                    print("rejected seq:", line)
                else:
                    output_file.write(line)
            f.close()
        output_file.close()
    break 
```
