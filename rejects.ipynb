```python
import os
```

```python
input_path = "C:/Users/maksi/Documents/Statistics/Projects/Avian_Flu_Files/Alignments/11-01-2021--06-27-2025_B3_13_D1_1/B3_13/"
genotype = "B3_13"
os.chdir(input_path)
reject_lines = []

# output_path = input_path + genotype + "_test.fasta" # Genotype and Segment should all be the same

for dirpath, dirs, files in os.walk(input_path + "saved/"):
    for file in files:
        file_name = os.path.join(dirpath, file)
        print(file_name)
        if "rejects" in file_name:
            with open(file_name) as r:
                lines = r.readlines()
                for i, line in enumerate(lines):
                    reject_lines.append(line)
                r.close()

print(reject_lines)
```

```python
for dirpath, dirs, files in os.walk(input_path):
    for file in files:
        file_name = os.path.join(dirpath, file)
        output_file = open(file_name[:-6] + "_edit.fasta", "w")
        with open(file_name) as f:
            lines = f.readlines()
            for line in lines:
                if line in reject_lines:
                    print("rejected:", line)
                else:
                    output_file.write(line)
            f.close()
        output_file.close()
    break 
```
