```python
import os

genotype = "B3.13"
start_date = "11-01-2021"
end_date = "07-04-2025"

date_range = start_date + "--" + end_date

input_path = "C:/Users/maksiaevai.NCBI_NT/OneDrive - National Institutes of Health/Documents/Virus_Evolution/Avian_Flu_Files/Alignments/11-01-2021--07-04-2025/example/trimmed/" # + genotype + "/trimmed/"

os.chdir(input_path)
sequences = {}
isolates = {}

# duplicates = {}
# output_path = input_path + "concat.fasta" # Genotype and Segment should all be the same


for dirpath, dirs, files in os.walk(input_path):
    counter = 0
    # current_header = ""
    
    for file in files:
        
        print(counter)
        file_name = os.path.join(dirpath, file)
        print(file_name)
        
        # sequences[header] = "" # Put header in list of headers to get sequences
        
        
        with open(file_name) as f:
            lines = f.readlines()
            header = ""
            isolate = ""
            for l in lines:
                # already_an_instance = False
                if l[0] == ">": # If header
                    header = l
                    isolate = header.split("|")[1]
                    if counter == 0: # and isolate in isolates.keys():

                        isolates[isolate] = header # Add isolate to list of seen isolates
                        sequences[header] = ""

                else: # If sequence
                    sequence = l
                    try:
                        sequences[header] = sequences[header] + sequence
                    except: # It's an NCBI Virus sequence whose header doesn't match the previous
                        if isolate in isolates.keys():
                            sequences[isolates[isolate]] = sequences[isolates[isolate]] + sequence
                        else:
                            print(header)
            f.close()
        counter += 1
    break 

print(sequences.keys())
print(len(sequences)) # There are different identifiers for each segment in NCBI Virus

# print(len(set(list(sequences.keys()))))

with open(input_path + genotype + "_concat_" + str(len(sequences)) + ".fasta", "w+") as o:
    for header in sequences.keys():
        # if header != "no":
        sequence = sequences[header]
        o.write(header + "\n")
        o.write(sequence + "\n")
    o.close()
```
