```python
import os
```

```python
genotype = "B3.7"
start_date = "11-01-2021"
end_date = "07-25-2025"

date_range = start_date + "--" + end_date

input_path = "C:/Users/maksiaevai.NCBI_NT/Documents/Avian_Flu_Files/Alignments/" + date_range + "/" + genotype + "/trimmed/"

os.chdir(input_path)
sequences = {}

output_path = input_path + "concat.fasta" # Genotype and Segment should all be the same

for dirpath, dirs, files in os.walk(input_path):
    for file in files:
        file_name = os.path.join(dirpath, file)
        print(file_name)
        header = ""
        with open(file_name) as f:
            lines = f.readlines()
            lines_list = []
            for line in lines:
                lines_list.append(line)
            for i, l in enumerate(lines_list):
                if i % 2 == 0: # If header
                    header = l
                    if header not in sequences:
                        sequences[header] = ""
                else: # If sequence
                    sequence = l
                    sequences[header] = sequences[header] + sequence
            f.close()

print(sequences)

with open(input_path + genotype + "_" + date_range + "_concat_" + str(len(sequences)) + ".fasta", "w+") as o:
    for header in sequences:
        sequence = sequences[header]
        o.write(header + "\n")
        o.write(sequence + "\n")
    o.close()
```
