```python
import os
from collections import defaultdict
```

```python
genotype = "D1_1"

input_path = "C:/Users/maksiaevai.NCBI_NT/Documents/Avian_Flu_Files/Alignments/11-01-2021--06-27-2025/" + genotype + "/trimmed/"

os.chdir(input_path)
sequences = defaultdict(list)

# output_path = input_path + genotype + "_test.fasta" # Genotype and Segment should all be the same

for dirpath, dirs, files in os.walk(input_path):
    for file in files:
        file_name = os.path.join(dirpath, file)
        print(file_name)
        header = ""
        with open(file_name) as f:
            lines = f.readlines()
            lines_list = []
            for line in lines:
                lines_list.append(line)
            for i, l in enumerate(lines_list):
                if i % 2 == 0: # If header
                    header = l
                else: # If sequence
                    sequence = l
                    sequences[header].append(sequence)
            f.close()
        # if "rejects" in file_name:
        #     with open(file_name) as r:
        #         lines = r.readlines()
        #         for i, line in enumerate(lines):
        #             reject_lines.append(line)
        #         r.close()

# print(reject_lines)
```
