```python
genotype = "Unassigned"
start_date = "11-01-2021"
end_date = "11-14-2025"

date_range = start_date + "--" + end_date

# input_path = "C:/Users/maksiaevai.NCBI_NT/Documents/Avian_Flu_Files/Alignments/" + date_range + "/" + genotype + "/trimmed/"
# input_path = "C:/Users/maksi/Documents/Statistics/Projects/Avian_Flu_Files/Combinations/NCBI_Virus_Andersen_GISAID/11-01-2021--09-19-2025_Antarctica_North_America_South_America/D1.1/"
# input_path = "C:/Users/maksi/Documents/Statistics/Projects/Avian_Flu_Data/2021-11-01--2025-10-03_Antarctica_South_America_B3_2/trimmed_alns/"
# input_path = "C:/Users/maksiaevai.NCBI_NT/Documents/Avian_Flu_Files/Alignments/" + date_range + "/" + genotype + "/"
input_path = "C:/Users/maksiaevai.NCBI_NT/Documents/Avian_Flu_Files/Combinations/NCBI_Virus_Andersen_GISAID/11-01-2021--11-14-2025_Antarctica_North_America_South_America/unassigned/"


os.chdir(input_path)
sequences = {}

# output_path = input_path + "concat.fasta" # Genotype and Segment should all be the same

for dirpath, dirs, files in os.walk(input_path):
    for file in files:
        file_name = os.path.join(dirpath, file)
        print(file_name)
        header = ""
        with open(file_name) as f:
            lines = f.readlines()
            lines_list = []
            for line in lines:
                lines_list.append(line)
            for i, l in enumerate(lines_list):
                if l[0] == ">": # If header
                    header = l
                    if header not in sequences.keys():
                        sequences[header] = ""
                else: # If sequence
                    sequence = l
                    sequences[header] = sequences[header] + sequence
            f.close()
    break 

print(sequences.keys())

with open(input_path + genotype + "_concat_" + str(len(sequences)) + ".fasta", "w+") as o:
    for header in sequences:
        sequence = sequences[header]
        o.write(header + "\n")
        o.write(sequence + "\n")
    o.close()
```
