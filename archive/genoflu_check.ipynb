```python
# Housekeeping

import os
import glob 
import pandas as pd

home = "C:/Users/maksi/Documents/Statistics/Projects/Avian_Flu_Files/"
sub_directory = "Other/"

os.chdir(home + sub_directory)
```

```python
fasta_files = {}
headers = []
sequences = []
for dirpath, dirs, files in os.walk(home + sub_directory):
    for n, file in enumerate(files):
        file_name = os.path.join(dirpath, file)
        if ".fasta" in file_name:
            with open(file_name) as f:
                lines = f.readlines()
                for num, line in enumerate(lines):
                    if line[0] == ">":
                        header = line[1:].strip()
                        headers.append(header)
                    else:
                        if num < len(lines):
                            i = num
                            sequence = "" 
                            while i < len(lines) - 1 and lines[i + 1][0] != ">":
                                sequence = sequence + lines[i + 1].strip()
                                i += 1
                            sequences.append(sequence)
            f.close()
                    
for num, header in enumerate(headers):
    fasta_files[header] = sequences[num]

print(headers)
print(fasta_files.keys())

```

```python
print(fasta_files["EPI_ISL_19752381|A/chicken/Chaco/516-4/2025|A_/_H5N1|HA|2025-02-06"])

# Create temporary files for GenoFlu
for header in fasta_files.keys():
    piece = header.split("|")[0]
    print(piece)
    output_path = home + sub_directory + piece + ".fasta" # Genotype and Segment should all be the same
    output_file = open(output_path, "w")
    output_file.write(">" + header + "\n")
    for lines in fasta_files[header]:
        
        output_file.write(lines)
    
    output_file.close()
```
