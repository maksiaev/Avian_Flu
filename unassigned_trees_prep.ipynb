```python
import os
import random
```

```python
start_date = "11-01-2021"
end_date = "11-28-2025"

input_path = "C:/Users/maksiaevai.NCBI_NT/Documents/Avian_Flu_Files/Combinations/NCBI_Virus_Andersen_GISAID/" + start_date + "--" + end_date + "_Antarctica_North_America_South_America/"

os.chdir(input_path)
```

```python
sequences = {}
genotypes = []
counter = 1
# output_path = input_path + "concat.fasta" # Genotype and Segment should all be the same

for dirpath, dirs, files in os.walk(input_path):
    for file in files:
        file_name = os.path.join(dirpath, file)
        print(file_name)
        genotype = "_".join(file_name.split("_")[-3:-1])
        if genotype not in genotypes:
            genotypes.append(genotype)
        header = ""
        with open(file_name) as f:
            lines = f.readlines()
            lines_list = []
            for line in lines:
                lines_list.append(line)
            for i, l in enumerate(lines_list):
                if l[0] == ">": # If header
                    header = l
                    if header not in sequences.keys():
                        sequences[header] = ""
                else: # If sequence
                    sequence = l
                    sequences[header] = sequences[header] + sequence
            f.close()
    # break 
print(genotypes)
# print(sequences.keys())
for genotype in genotypes:
    if "Unassigned" not in genotype:
        genotype_sequences = {}
        for header in sequences:
            if genotype.split("_")[0] in header:
                genotype_sequences[header] = sequences[header]

        output_path = "C:/Users/maksiaevai.NCBI_NT/Documents/Avian_Flu_Files/Alignments/"
        random_genotype_sequences = {}
        for i in range(10):
            if len(genotype_sequences) == 0:
                break 
            header = random.choice(list(genotype_sequences.keys()))
            random_genotype_sequences[header] = genotype_sequences[header]
            genotype_sequences.pop(header)
        # for genotype in genotypes:
        with open(output_path + genotype + "_random_ten_concat.fasta", "w+") as o:
            for header in random_genotype_sequences:
                if genotype in header:
                    sequence = random_genotype_sequences[header]
                    o.write(header + "\n")
                    o.write(sequence + "\n")
            o.close()
    else:
        with open(output_path + genotype + "_concat.fasta", "w+") as o2:
            for header in sequences:
                if "Not_assigned" in header:
                    print(header)
                    sequence = sequences[header]
                    o2.write(header + "\n")
                    o2.write(sequence + "\n")
            o2.close()
```
