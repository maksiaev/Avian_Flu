```python
import os
import random
```

```python
start_date = "11-01-2021"
end_date = "12-05-2025"

input_path = "C:/Users/maksiaevai.NCBI_NT/Documents/Avian_Flu_Files/Combinations/NCBI_Virus_Andersen_GISAID/" + start_date + "--" + end_date + "_Antarctica_North_America_South_America/"

os.chdir(input_path)
```

```python
# Merge all files into 8 segments

random.seed(10)

output_path = input_path + "mega_segment_"

for dirpath, dirs, files in os.walk(input_path):
    for file in files:
        file_name = os.path.join(dirpath, file)
        print(file_name)
        segment = file_name.split("_")[-2] # Get segment name
        genotype = file_name.split("_")[-3] # Get genotype
        # print(segment)
        # break 
        sequences = {}
        random_sequences = {}
        with open(file_name) as f:
            line_list = []
            lines = f.readlines()
            for line in lines:
                line_list.append(line)
            for i, l in enumerate(line_list):
                if l[0] == ">": # If header
                    header = l
                    if header not in sequences.keys():
                        sequences[header] = ""
                else: # If sequence
                    sequence = l
                    sequences[header] = sequences[header] + sequence
            for i in range(10): # If assigned, add only up to 10 random 
                if len(sequences) == 0:
                    break 
                if genotype != "Unassigned":
                    header = random.choice(list(sequences.keys()))
                    random_sequences[header] = sequences[header]
                    sequences.pop(header)
            if genotype != "Unassigned":
                with open(output_path + segment + ".fasta", "a+") as o: # Append lines to output file
                    for header in random_sequences.keys():
                        o.write(header + "\n")
                        o.write(random_sequences[header] + "\n")
                    o.close()
            else: # If unassigned, add all
                with open(output_path + segment + ".fasta", "a+") as o: # Append lines to output file
                    for header in sequences.keys():
                        o.write(header + "\n")
                        o.write(sequences[header] + "\n")
                    o.close()
```
